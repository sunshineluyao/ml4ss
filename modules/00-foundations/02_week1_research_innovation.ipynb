**`PROXY_NOISE` — Standard deviation of error added to the simulated need measure**

- **Default:** `0.8`
- **What you may enter:** 0.2 to 2.0, in steps of 0.2
- **Example and expected effect:** Try `0.2, then 2.0`. Usually increases proxy scatter and measurement error when raised; downstream results can change.


**`BUDGET_SHARE` — Share of 600 simulated people who can receive support**

- **Default:** `0.3`
- **What you may enter:** 0.1 to 0.8, in steps of 0.1
- **Example and expected effect:** Try `0.2, then 0.5`. Changes allocation capacity from 120 to 300 people in this example. It does not change the randomized experiment.


**`SEED` — Seed controlling simulated data, splits, and randomized model components**

- **Default:** `201`
- **What you may enter:** An integer from 0 to 4294967294
- **Example and expected effect:** Try `202`. Changes the simulated sample and random components; hold fixed when comparing another parameter.


### Try a controlled change

- **Measurement:** keep `SEED = 201`; compare `PROXY_NOISE = 0.2` with `2.0`. Expect a wider proxy scatter and larger measurement error at high noise. Predictive performance may change; inspect the actual result rather than promising every score will move monotonically.
- **Allocation:** keep the seed and noise fixed; compare `BUDGET_SHARE = 0.2` with `0.5`. Expect capacity to change from 120 to 300 people. Compare the total-population average gain of all three rules at the same capacity. These gains use the treatment benefits specified by this simulation.
- **Replication:** change `SEED` from 201 to 202 only after the controlled comparison. Expect a different random sample and estimates; a single favorable run does not establish a general result.

**Scope check:** selecting `PURPOSE = "Causal inference"` records that choice. All methods still run. Writing "Kunshan" in `CONTEXT` records your setting; it does not turn the simulated people into observed Kunshan data.

````python
# @title ✏️ STUDENT INPUT — Part 1 settings
GROUP_LABEL = "Group 01" # @param {type:"string"}
MEMBER_COUNT = 2 # @param {type:"slider", min:1, max:3, step:1}
CONTEXT = "A community or setting we want to understand" # @param {type:"string"}
PURPOSE = "Prediction" # @param ["Measurement", "Explanation", "Prediction", "Causal inference", "Optimization", "Frontier methods"]
PROXY_NOISE = 0.8 # @param {type:"slider", min:0.2, max:2.0, step:0.2}
BUDGET_SHARE = 0.30 # @param {type:"slider", min:0.1, max:0.8, step:0.1}
SEED = 201 # @param {type:"integer"}

assert isinstance(GROUP_LABEL, str) and GROUP_LABEL.strip(), 'Enter a group or individual label.'
assert MEMBER_COUNT in (1, 2, 3), 'Use one, two, or three participants.'
assert PURPOSE in ['Measurement','Explanation','Prediction','Causal inference','Optimization','Frontier methods']
assert 0.2 <= PROXY_NOISE <= 2.0, 'Proxy noise must be between 0.2 and 2.0.'
assert 0.1 <= BUDGET_SHARE <= 0.8, 'Budget share must be between 0.1 and 0.8.'
assert isinstance(SEED, int) and 0 <= SEED <= 4294967294, 'Use an integer seed from 0 to 4294967294.'
print(f'{GROUP_LABEL}: {MEMBER_COUNT} participant(s); intended purpose: {PURPOSE}')
print(f'Simulation: 600 people; proxy-noise SD = {PROXY_NOISE}; support capacity = {int(BUDGET_SHARE * 600)}; seed = {SEED}')
print('PURPOSE and CONTEXT record your reasoning; all numerical demonstrations run.')
````

## 1.5 Prepare the supplied analysis code

▶️ **RUN & OBSERVE; no implementation edits needed.** Run the next cell once. It imports the packages, creates an output folder, and defines the supplied analysis and inline-review functions. Expand **Show code** if you want to inspect how a chart or calculation is made.

````python
# @title ▶️ RUN — Load packages and supplied analysis functions
"""Self-contained, CPU-only support code embedded in the teaching notebook."""
import os, sys, re, io, json, time, math, hashlib, html, platform
from pathlib import Path
from datetime import datetime, timezone
import urllib.request, urllib.parse, urllib.error
import xml.etree.ElementTree as ET
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.inspection import permutation_importance
from IPython.display import display, HTML

COLORS=['#168F83','#DF6A62','#7C68B5','#D49B31','#3977A5','#526579']
plt.rcParams.update({'figure.dpi':120,'axes.spines.top':False,'axes.spines.right':False,'axes.titleweight':'bold','font.size':10})
RUN_STAMP=datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
OUT=Path('ml4ss_results')/f'investigation_{RUN_STAMP}'
OUT.mkdir(parents=True,exist_ok=True)
RUN={'notebook_revision':'2026-10-26-three-part-guide','context':CONTEXT,'purpose':PURPOSE,'members':MEMBER_COUNT,'seed':SEED,'started_utc':RUN_STAMP,'query_receipts':[]}
FIGURES=[]
LAST_ARXIV_REQUEST=0.0

def save_figure(fig,name,show=True):
    fig.savefig(OUT/f'{name}.png',bbox_inches='tight',facecolor='white')
    fig.savefig(OUT/f'{name}.pdf',bbox_inches='tight',facecolor='white')
    FIGURES.append((name,OUT/f'{name}.png'))
    if show:display(fig)
    plt.close(fig)

def get_json(url,params,timeout=9):
    full=url+'?'+urllib.parse.urlencode(params)
    req=urllib.request.Request(full,headers={'User-Agent':'ML4SS-Teaching-Lab/1.0 (public metadata research)'})
    with urllib.request.urlopen(req,timeout=timeout) as r:
        if r.status!=200:raise RuntimeError(f'HTTP {r.status}')
        payload=json.load(r)
    RUN['query_receipts'].append({'url':full,'retrieved_utc':datetime.now(timezone.utc).isoformat(),'status':'observed'})
    return payload

def unroll_abstract(inv):
    seq={i:w for w,pos in (inv or {}).items() for i in pos}
    return ' '.join(seq[i] for i in sorted(seq))

def oa_records(payload,query):
    rows=[]
    for w in payload.get('results',[]):
        loc=w.get('primary_location') or {}
        rows.append({'id':w['id'],'doi':w.get('doi') or '', 'title':w.get('display_name') or '', 'year':w.get('publication_year'),'authors':'; '.join(a['author']['display_name'] for a in w.get('authorships',[])[:8]),'venue':(loc.get('source') or {}).get('display_name',''),'abstract':unroll_abstract(w.get('abstract_inverted_index')),'cited_by_count':w.get('cited_by_count',0),'url':w.get('doi') or w['id'],'search_topic':query})
    return rows

def retrieve(query):
    """Never substitute a different topic when a query has no cached matches."""
    errors=[]
    if RETRIEVAL_MODE!='snapshot':
        try:
            p={'search':query,'filter':f'publication_year:{START_YEAR}-{END_YEAR}','per_page':min(int(MAX_WORKS),100)}
            raw=get_json('https://api.openalex.org/works',p)
            rows=oa_records(raw,query)
            if not rows:raise ValueError('OpenAlex returned no works')
            return rows,'live OpenAlex',raw.get('meta',{}).get('count'),errors
        except Exception as e:errors.append('OpenAlex: '+type(e).__name__)
        try:
            raw=get_json('https://api.crossref.org/works',{'query':query,'rows':min(int(MAX_WORKS),100),'filter':f'from-pub-date:{START_YEAR}-01-01,until-pub-date:{END_YEAR}-12-31'})
            rows=[]
            for w in raw.get('message',{}).get('items',[]):
                dp=(w.get('published') or w.get('issued') or {}).get('date-parts',[[None]])
                rows.append({'id':'https://doi.org/'+w.get('DOI',''),'doi':'https://doi.org/'+w.get('DOI',''),'title':' '.join(w.get('title',[])),'year':dp[0][0],'authors':'; '.join(' '.join([a.get('given',''),a.get('family','')]).strip() for a in w.get('author',[])[:8]),'venue':'; '.join(w.get('container-title',[])),'abstract':re.sub('<[^>]+>',' ',w.get('abstract','')),'cited_by_count':w.get('is-referenced-by-count',0),'url':w.get('URL',''),'search_topic':query})
            if rows:return rows,'live Crossref',None,errors
        except Exception as e:errors.append('Crossref: '+type(e).__name__)
    exact=next((t for t in SNAPSHOT['topics'] if t['topic'].casefold()==query.strip().casefold()),None)
    if exact:
        rows=[r.copy() for r in exact.get('records',[]) if r.get('year') and START_YEAR<=r['year']<=END_YEAR][:MAX_WORKS]
    else:
        terms=re.findall(r'\w+',query.casefold())
        pool={r['id']:r for t in SNAPSHOT['topics'] for r in t.get('records',[])}
        rows=[r.copy() for r in pool.values() if terms and all(x in (r.get('title','')+' '+r.get('abstract','')).casefold() for x in terms) and r.get('year') and START_YEAR<=r['year']<=END_YEAR][:MAX_WORKS]
    RUN['query_receipts'].append({'query':query,'status':'snapshot fallback','snapshot_source':'OpenAlex metadata','snapshot_dates':[t.get('retrieved_at') for t in SNAPSHOT['topics']],'live_errors':errors,'exact_preset':exact is not None})
    return rows,'saved OpenAlex sample',None,errors

METHOD_RULES={
 'Measurement':['measurement','measure','text analysis','sentiment','remote sensing','survey','classification','mapping'],
 'Explanation':['explain','hypothesis','mechanism','interpretab','theory','association','correlation'],
 'Prediction':['predict','forecast','out-of-sample','supervised','regression','machine learning'],
 'Causal inference':['causal','randomized','randomised','treatment effect','difference-in-differences','instrumental variable','experiment'],
 'Optimization':['optimiz','optimis','allocation','policy learning','targeting','reinforcement learning','decision'],
 'Frontier methods':['foundation model','large language model','generative ai','transformer','agentic','in-context']}

def prepare_records(rows):
    cols=['id','doi','title','year','authors','venue','abstract','cited_by_count','url','search_topic']
    df=pd.DataFrame(rows,columns=cols).drop_duplicates(subset=['id']).copy()
    df['year']=pd.to_numeric(df['year'],errors='coerce')
    df['abstract_available']=df['abstract'].fillna('').str.len().gt(0)
    for method,words in METHOD_RULES.items():
        df[method]=[(any(w in (str(t)+' '+str(a)).lower() for w in words)) for t,a in zip(df['title'],df['abstract'])]
    df['method_status']='Unverified keyword screening, not an assessment of the paper'
    return df

def retrieve_arxiv(query):
    global LAST_ARXIV_REQUEST
    tokens=re.findall(r'[\w-]+',query)[:14]
    if not tokens:return [],None,'No valid query tokens'
    expression=' AND '.join('all:'+t for t in tokens)
    expression+=f' AND submittedDate:[{START_YEAR}01010000 TO {END_YEAR}12312359]'
    params={'search_query':expression,'start':0,'max_results':min(MAX_WORKS,100),'sortBy':'relevance','sortOrder':'descending'}
    url='https://export.arxiv.org/api/query?'+urllib.parse.urlencode(params)
    delay=3.1-(time.monotonic()-LAST_ARXIV_REQUEST)
    if delay>0:time.sleep(delay)
    LAST_ARXIV_REQUEST=time.monotonic()
    req=urllib.request.Request(url,headers={'User-Agent':'ML4SS-Teaching-Lab/1.0 (arXiv public-metadata discovery)'})
    with urllib.request.urlopen(req,timeout=10) as response:root=ET.fromstring(response.read())
    ns={'a':'http://www.w3.org/2005/Atom','o':'http://a9.com/-/spec/opensearch/1.1/','x':'http://arxiv.org/schemas/atom'}
    total=root.findtext('o:totalResults',default='0',namespaces=ns)
    rows=[]
    for e in root.findall('a:entry',ns):
        ident=e.findtext('a:id','',ns).replace('http://','https://')
        if '/abs/' not in ident:continue
        doi=e.findtext('x:doi','',ns)
        published=e.findtext('a:published','',ns)
        rows.append({'id':ident,'doi':'https://doi.org/'+doi if doi else '', 'title':' '.join(e.findtext('a:title','',ns).split()),'year':int(published[:4]) if published[:4].isdigit() else None,'authors':'; '.join(x.findtext('a:name','',ns) for x in e.findall('a:author',ns)),'venue':e.findtext('x:journal_ref','arXiv preprint; verify publication status',ns),'abstract':' '.join(e.findtext('a:summary','',ns).split()),'cited_by_count':None,'url':ident,'search_topic':query})
    RUN['query_receipts'].append({'url':url,'retrieved_utc':datetime.now(timezone.utc).isoformat(),'status':'observed','provider':'arXiv','query_semantics':'all-field AND terms plus submitted-date range'})
    return rows,int(total),'Live arXiv Atom API'

def retrieve_crossref(query):
    raw=get_json('https://api.crossref.org/works',{'query':query,'rows':min(MAX_WORKS,100),'filter':f'from-pub-date:{START_YEAR}-01-01,until-pub-date:{END_YEAR}-12-31'})
    rows=[]
    for w in raw.get('message',{}).get('items',[]):
        dp=(w.get('published') or w.get('issued') or {}).get('date-parts',[[None]])
        rows.append({'id':'https://doi.org/'+w.get('DOI',''),'doi':'https://doi.org/'+w.get('DOI',''),'title':' '.join(w.get('title',[])),'year':dp[0][0],'authors':'; '.join(' '.join([a.get('given',''),a.get('family','')]).strip() for a in w.get('author',[])[:8]),'venue':'; '.join(w.get('container-title',[])),'abstract':re.sub('<[^>]+>',' ',w.get('abstract','')),'cited_by_count':w.get('is-referenced-by-count',0),'url':w.get('URL',''),'search_topic':query})
    return rows,raw.get('message',{}).get('total-results'),'Live Crossref REST API'

def compare_sources(query,primary_rows,primary_route):
    """Compare observed retrievals. Unavailable services are missing, never zero."""
    provider_rows={};summary=[]
    if 'OpenAlex' in primary_route:
        provider_rows['OpenAlex']=primary_rows
        summary.append({'provider':'OpenAlex','status':primary_route,'retrieved_n':len(primary_rows),'search_total':total_matching,'query_semantics':'OpenAlex search relevance; publication-year range'})
    else:
        provider_rows['Crossref']=primary_rows
        summary.append({'provider':'Crossref','status':primary_route,'retrieved_n':len(primary_rows),'search_total':None,'query_semantics':'Crossref query relevance; publication-date range'})
    for name,fun,semantics in [('arXiv',retrieve_arxiv,'AND terms across all fields; submitted-date range'),('Crossref',retrieve_crossref,'Query relevance; publication-date range')]:
        if name in provider_rows:continue
        if RETRIEVAL_MODE=='snapshot':
            summary.append({'provider':name,'status':'Not queried in snapshot mode','retrieved_n':None,'search_total':None,'query_semantics':semantics});continue
        try:
            rows,total,status=fun(query);provider_rows[name]=rows
            summary.append({'provider':name,'status':status,'retrieved_n':len(rows),'search_total':total,'query_semantics':semantics})
        except Exception as e:
            # Cached arXiv reference metadata can support source inspection, but is not a live query.
            cached=ARXIV_SNAPSHOT.get(query.casefold()) if 'ARXIV_SNAPSHOT' in globals() and name=='arXiv' else None
            if cached and cached.get('records'):
                rows=[r for r in cached['records'] if isinstance(r.get('year'),int) and START_YEAR<=r['year']<=END_YEAR][:MAX_WORKS];provider_rows[name]=rows
                summary.append({'provider':name,'status':'Saved arXiv query sample; live '+type(e).__name__,'retrieved_n':len(rows),'search_total':None,'query_semantics':cached.get('query_semantics',semantics)})
            else:summary.append({'provider':name,'status':'Unavailable: '+type(e).__name__,'retrieved_n':None,'search_total':None,'query_semantics':semantics})
            RUN['query_receipts'].append({'provider':name,'query':query,'status':'unavailable','error_type':type(e).__name__})
    aliases={};merged=[];keys_by_provider={}
    for provider,rows in provider_rows.items():
        ids=set()
        for record in rows:
            d=str(record.get('doi') or '').lower().removeprefix('https://doi.org/').strip()
            ident=str(record.get('id',''));arxiv=re.search(r'arxiv\.org/abs/([^?#]+)',ident)
            title=re.sub(r'\W+','',str(record.get('title','')).casefold())
            keys=[]
            if d:keys.append('doi:'+d)
            if arxiv:keys.append('arxiv:'+re.sub(r'v\d+$','',arxiv.group(1)))
            if title:keys.append('titleyear:'+title+':'+str(record.get('year')))
            if not keys:continue
            found=next((aliases[k] for k in keys if k in aliases),None)
            if found is None:
                found=len(merged);entry=record.copy();entry['source_providers']=provider;entry['source_ids']=ident;merged.append(entry)
            else:
                entry=merged[found]
                entry['source_providers']='; '.join(sorted(set(entry['source_providers'].split('; ')+[provider])))
                entry['source_ids']+='; '+ident
            for k in keys:aliases[k]=found
            ids.add(found)
        keys_by_provider[provider]=ids
    table=pd.DataFrame(summary)
    for name in ['OpenAlex','arXiv','Crossref']:
        if name not in table.provider.values:table.loc[len(table)]={'provider':name,'status':'Not available in this run','retrieved_n':None,'search_total':None,'query_semantics':'Not queried'}
    table['note']='Sample sizes are comparable retrieval diagnostics; query totals and peer-review status are not equivalent.'
    for provider,rows in provider_rows.items():
        ix=table.index[table.provider.eq(provider)]
        table.loc[ix,'abstract_share']=np.mean([bool(str(r.get('abstract') or '').strip()) for r in rows]) if rows else np.nan
        table.loc[ix,'DOI_share']=np.mean([bool(str(r.get('doi') or '').strip()) for r in rows]) if rows else np.nan
        other=set().union(*(v for k,v in keys_by_provider.items() if k!=provider))
        table.loc[ix,'unique_within_this_run']=len(keys_by_provider[provider]-other)
    table.to_csv(OUT/'source_comparison.csv',index=False)
    pd.DataFrame(merged).to_csv(OUT/'combined_sources_deduplicated.csv',index=False)
    names=list(keys_by_provider);overlap=pd.DataFrame(index=names,columns=names,dtype=float)
    for a in names:
        for b in names:
            union=keys_by_provider[a]|keys_by_provider[b]
            overlap.loc[a,b]=len(keys_by_provider[a]&keys_by_provider[b])/len(union) if union else np.nan
    overlap.to_csv(OUT/'source_overlap_jaccard.csv')
    usable=table.dropna(subset=['retrieved_n'])
    fig,axs=plt.subplots(1,2,figsize=(10,4.1))
    axs[0].bar(usable.provider,usable.retrieved_n,color=COLORS[:len(usable)]);axs[0].set(ylabel='Records returned',title='A capped sample from each source')
    x=np.arange(len(usable));axs[1].bar(x-.18,usable.abstract_share*100,width=.36,color=COLORS[0],label='Abstract present');axs[1].bar(x+.18,usable.DOI_share*100,width=.36,color=COLORS[1],label='DOI present');axs[1].set_xticks(x,usable.provider);axs[1].set(ylabel='Percent of retrieved records',ylim=(0,110),title='Different metadata for the same topic');axs[1].legend(fontsize=8)
    fig.suptitle('How does the source shape what you can inspect?',weight='bold');fig.tight_layout()
    fig.text(.01,-.03,'Unavailable sources are missing. Equal sample caps do not imply equal database coverage.',fontsize=9)
    save_figure(fig,'12_source_comparison')
    RUN['source_comparison']=table.where(pd.notna(table),None).to_dict('records')
    return table,pd.DataFrame(merged),overlap

def simulate(n=600):
    rng=np.random.default_rng(SEED)
    need=rng.normal(size=n)
    access=rng.normal(size=n)
    proxy=need+rng.normal(0,PROXY_NOISE,size=n)
    y0=10+2*need+1.5*access+1.1*np.sin(1.6*need)+rng.normal(0,1.5,size=n)
    tau=0.7+2*np.exp(-0.5*need**2)
    treatment=rng.binomial(1,0.5,size=n)
    observed=y0+treatment*tau
    return pd.DataFrame({'need_latent':need,'access':access,'need_proxy':proxy,'treatment':treatment,'y0':y0,'tau':tau,'outcome':observed})

def methods_demo():
    d=simulate();d.to_csv(OUT/'synthetic_teaching_data.csv',index=False)
    fig,ax=plt.subplots(figsize=(7.4,4.2))
    ax.scatter(d.need_latent,d.need_proxy,s=13,alpha=.38,color=COLORS[0])
    ax.plot([-3,3],[-3,3],color=COLORS[1],ls='--',label='Perfect measurement')
    corr=float(d[['need_latent','need_proxy']].corr().iloc[0,1]);mse=float(np.mean((d.need_proxy-d.need_latent)**2))
    ax.set(xlabel='Known latent construct in simulation',ylabel='Noisy observed proxy',title=f'Measurement: proxy noise = {PROXY_NOISE:.2f}')
    ax.legend();save_figure(fig,'01_measurement')
    X=d[['need_proxy','access','treatment']];y=d.outcome
    Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.30,random_state=SEED)
    models={'Mean baseline':DummyRegressor(),'Linear regression':LinearRegression(),'Random forest':RandomForestRegressor(n_estimators=80,max_depth=5,min_samples_leaf=8,random_state=SEED,n_jobs=1)}
    scores=[]
    for name,m in models.items():
        m.fit(Xtr,ytr);pred=m.predict(Xte)
        scores.append({'model':name,'test_RMSE':float(np.sqrt(mean_squared_error(yte,pred))),'test_R2':float(r2_score(yte,pred))})
    scores=pd.DataFrame(scores);scores.to_csv(OUT/'prediction_comparison.csv',index=False)
    linear=models['Linear regression'];coefs=pd.DataFrame({'feature':X.columns,'coefficient':linear.coef_})
    coefs.to_csv(OUT/'explanation_coefficients.csv',index=False)
    fig,ax=plt.subplots(figsize=(7.4,4.2));ax.barh(coefs.feature,coefs.coefficient,color=COLORS[:3]);ax.axvline(0,color='#526579',lw=1)
    ax.set(xlabel='Coefficient conditional on the included variables',title='Explanation: what does this fitted model associate with outcomes?')
    save_figure(fig,'02_explanation')
    fig,ax=plt.subplots(figsize=(7.4,4.2));ax.bar(scores.model,scores.test_RMSE,color=COLORS[:3]);ax.set(ylabel='Held-out root mean squared error',title='Prediction: compare all models on the same held-out observations')
    ax.tick_params(axis='x',labelrotation=10);save_figure(fig,'03_prediction')
    treated=d.loc[d.treatment==1,'outcome'];control=d.loc[d.treatment==0,'outcome']
    effect=float(treated.mean()-control.mean());se=float(np.sqrt(treated.var(ddof=1)/len(treated)+control.var(ddof=1)/len(control)))
    causal=pd.DataFrame([{'estimand':'Average treatment effect','estimated_randomized_difference':effect,'SE':se,'CI_low':effect-1.96*se,'CI_high':effect+1.96*se,'simulation_truth':float(d.tau.mean())}])
    causal.to_csv(OUT/'causal_estimate.csv',index=False)
    fig,ax=plt.subplots(figsize=(7.4,4.2));ax.errorbar([effect],[0],xerr=[1.96*se],fmt='o',color=COLORS[0],capsize=6,label='Randomized estimate and approximate 95% CI');ax.axvline(d.tau.mean(),color=COLORS[1],ls='--',label='Known simulation truth');ax.axvline(0,color='#aaa',lw=1)
    ax.set(xlabel='Outcome units',yticks=[],title='Causal inference: what changed because of treatment?');ax.legend(loc='upper left',fontsize=8);save_figure(fig,'04_causal_inference')
    # A known-effect oracle is an upper reference inside this simulation only.
    k=max(1,int(BUDGET_SHARE*len(d)))
    random_order=np.random.default_rng(SEED+1).permutation(len(d));need_order=np.argsort(-d.need_latent.to_numpy());oracle_order=np.argsort(-d.tau.to_numpy())
    policy=[]
    for name,order in [('Random allocation',random_order),('Highest measured need',np.argsort(-d.need_proxy.to_numpy())),('Known-effect oracle',oracle_order)]:
        chosen=np.zeros(len(d),dtype=int);chosen[order[:k]]=1
        policy.append({'policy':name,'people_selected':int(chosen.sum()),'budget_share':BUDGET_SHARE,'mean_gain_per_person':float(np.mean(chosen*d.tau))})
    policy=pd.DataFrame(policy);policy.to_csv(OUT/'allocation_comparison.csv',index=False)
    fig,ax=plt.subplots(figsize=(7.4,4.2));ax.barh(policy.policy,policy.mean_gain_per_person,color=COLORS[:3]);ax.set(xlabel='Known simulated gain averaged over everyone',title=f'Optimization: whom to support with capacity for {k} of {len(d)} people?')
    save_figure(fig,'05_optimization')
    metrics={'measurement_proxy_correlation':corr,'measurement_MSE':mse,'causal_estimate':effect,'causal_truth':float(d.tau.mean()),'randomized_treatment':True,'synthetic':True,'split_seed':SEED,'test_size':.30,'n':len(d)}
    (OUT/'simulation_metrics.json').write_text(json.dumps(metrics,indent=2))
    return d,scores,coefs,causal,policy,metrics

def literature_gallery(df,mode):
    df.to_csv(OUT/'literature_records.csv',index=False)
    counts=df['year'].value_counts().sort_index() if len(df) else pd.Series(dtype=int)
    counts.rename_axis('year').rename('retrieved_sample_count').to_csv(OUT/'retrieved_year_counts.csv')
    fig,ax=plt.subplots(figsize=(8.2,4.2));ax.bar(counts.index,counts.values,color=COLORS[0]);ax.set(xlabel='Publication year',ylabel='Works in the retrieved sample',title=f'Your retrieved literature sample: {TOPIC}')
    if END_YEAR>=datetime.now(timezone.utc).year:ax.text(.01,.96,'Current-year records are incomplete',transform=ax.transAxes,va='top',fontsize=9,color=COLORS[1])
    save_figure(fig,'06_literature_sample_years')
    methodcounts=df[list(METHOD_RULES)].sum().astype(int) if len(df) else pd.Series(0,index=METHOD_RULES)
    methodcounts.rename('keyword_matches').to_csv(OUT/'method_screening_counts.csv')
    fig,ax=plt.subplots(figsize=(8.2,4.6));ax.barh(methodcounts.index,methodcounts.values,color=COLORS);ax.set(xlabel='Works matching at least one screening term',title='Method signals to verify, based on titles and available abstracts')
    ax.text(0,-.24,'Multiple labels are possible. Missing labels do not establish absence.',transform=ax.transAxes,fontsize=9)
    save_figure(fig,'07_method_screening')
    mat=pd.crosstab(df['year'],df['venue']) if len(df) else pd.DataFrame()
    # A readable heatmap of year and method, with no invented cross-paper edges.
    grid=df.groupby('year')[list(METHOD_RULES)].sum() if len(df) else pd.DataFrame(0,index=[START_YEAR],columns=list(METHOD_RULES))
    fig,ax=plt.subplots(figsize=(9,4.3));im=ax.imshow(grid.T.values,aspect='auto',cmap='YlGnBu');ax.set_xticks(range(len(grid.index)),[str(int(x)) for x in grid.index],rotation=45);ax.set_yticks(range(len(METHOD_RULES)),list(METHOD_RULES));ax.set_title('Where do screening signals appear within this sample?');fig.colorbar(im,ax=ax,label='Retrieved works with a keyword signal')
    save_figure(fig,'08_method_year_heatmap')
    return counts,methodcounts

def global_trend(query):
    """Return database-wide search counts separately from a ranked sample."""
    if RETRIEVAL_MODE=='snapshot':return None,'Not queried in snapshot mode'
    try:
        raw=get_json('https://api.openalex.org/works',{'search':query,'filter':f'publication_year:{START_YEAR}-{END_YEAR}','group_by':'publication_year','per_page':100})
        rows=[{'year':int(g['key']),'indexed_matching_works':int(g['count'])} for g in raw.get('group_by',[]) if str(g.get('key','')).isdigit()]
        trend=pd.DataFrame(rows).sort_values('year') if rows else pd.DataFrame(columns=['year','indexed_matching_works'])
        return trend,'Live OpenAlex annual search counts'
    except Exception as e:return None,'Unavailable: '+type(e).__name__

def frontier_demo(Xtr,Xte,ytr,yte):
    """An executable modest model comparison, never mislabeled as TabICL training."""
    result=[]
    for size in [40,100,len(Xtr)]:
        for name,model in [('Ridge',Ridge(alpha=1.0)),('Random forest',RandomForestRegressor(n_estimators=50,max_depth=5,min_samples_leaf=5,random_state=SEED))]:
            model.fit(Xtr.iloc[:size],ytr.iloc[:size]);pred=model.predict(Xte)
            result.append({'training_rows':size,'model':name,'test_RMSE':float(np.sqrt(mean_squared_error(yte,pred)))})
    frame=pd.DataFrame(result);frame.to_csv(OUT/'frontier_comparison_design.csv',index=False)
    fig,ax=plt.subplots(figsize=(7.4,4.2))
    for i,(name,g) in enumerate(frame.groupby('model')):ax.plot(g.training_rows,g.test_RMSE,marker='o',color=COLORS[i],label=name)
    ax.set(xlabel='Available training observations',ylabel='Held-out RMSE',title='A comparison design for evaluating a newer method');ax.legend();save_figure(fig,'09_frontier_evaluation_design')
    return frame

def original_3d_animation():
    from mpl_toolkits.mplot3d import Axes3D
    fig=plt.figure(figsize=(5,3),dpi=72);ax=fig.add_subplot(111,projection='3d')
    x=np.linspace(-2,2,25);a,b=np.meshgrid(x,x);c=(a-.5)**2+.5*(b+.4)**2
    ax.plot_surface(a,b,c,cmap='viridis',alpha=.8,linewidth=0)
    ax.set(xlabel='Choice 1',ylabel='Choice 2',zlabel='Specified loss',title='A conceptual optimization landscape')
    def update(frame):ax.view_init(elev=26,azim=frame*4);return []
    anim=FuncAnimation(fig,update,frames=24,interval=120)
    markup=anim.to_jshtml(default_mode='loop');(OUT/'10_conceptual_3d_animation.html').write_text(markup)
    plt.close(fig);return HTML(markup)

def show_response_block(title, responses):
    """Print complete, copyable responses in the notebook output."""
    import textwrap
    print(title)
    print()
    for label, value in responses.items():
        print(label.replace('_', ' ') + ':')
        text = str(value).strip() or 'Not yet written — complete this field in your own words.'
        for paragraph in text.splitlines():
            print(textwrap.fill(paragraph, width=100, replace_whitespace=False) if paragraph else '')
        print()

def part3_draft_status(responses):
    examples = [key for key, value in responses.items() if str(value).strip().lower().startswith('example')]
    blanks = [key for key, value in responses.items() if not str(value).strip()]
    if examples:
        print(f'Teaching-example fields remaining: {len(examples)}. Replace these with your own checked work.')
        print(', '.join(examples))
    if blanks:
        print('Incomplete fields: ' + ', '.join(blanks))
    print('Displayed text is a draft for human review. A completed run does not certify source checks, originality, or learning.')
    return {'example_fields': examples, 'blank_fields': blanks}

def show_investigation_report(df, scores, causal, policy, metrics, trend_status):
    """Display the current investigation in Colab; no file export or download."""
    configuration = {name: globals()[name] for name in ['GROUP_LABEL', 'MEMBER_COUNT', 'CONTEXT', 'PURPOSE', 'PROXY_NOISE', 'BUDGET_SHARE', 'SEED', 'TOPIC', 'COMPARISON_TOPIC', 'START_YEAR', 'END_YEAR', 'MAX_WORKS', 'RETRIEVAL_MODE']}
    answers = {'Q1_research_question': QUESTION_DRAFT, 'Q2_purpose_and_evidence': PURPOSE_EVIDENCE,
               'Q3_initial_expectation': INITIAL_EXPECTATION, 'Q4_supported_finding': SUPPORTED_FINDING,
               'Q5_revised_question': REVISED_QUESTION, 'Q6_next_evidence': NEXT_EVIDENCE,
               'AI_use_statement': AI_USE_STATEMENT}
    canvas = {name: globals()[name] for name in ['INNOVATION_ROUTE', 'CLOSEST_STUDIES', 'CANDIDATE_GAP', 'BASELINE_AND_TEST', 'DATA_AND_EVALUATION', 'CONTRIBUTIONS', 'PEER_FEEDBACK_REVISION', 'NEXT_WEEK_ACTION']}
    audit = {name: globals()[name] for name in ['MANUAL_CHECK_STATUS', 'MANUAL_PLATFORM', 'MANUAL_QUERY', 'MANUAL_DATE', 'ADDED_SOURCE', 'MISSED_REASON', 'SEARCH_REVISION', 'AUDIT_TERMS', 'BACKWARD_CHECK', 'FORWARD_CHECK', 'VERSION_CHECK']}
    checks = [{field: globals()[f'SOURCE_{i}_{field}'] for field in ['STATUS', 'TITLE', 'URL', 'CLAIM', 'SECTION', 'DECISION', 'REASON']} for i in [1, 2]]
    design = {name: globals()[name] for name in ['RESEARCH_PURPOSE', 'EVIDENCE_ROUTE', 'OBSERVATION_UNIT', 'EMPIRICAL_SOURCE', 'ML_TASK', 'BASELINE_PLAN', 'VALIDATION_PLAN', 'BOUNDED_CLAIM', 'ACCESS_AND_REVIEW', 'MONDAY_REVISION']}
    print('YOUR INVESTIGATION — INLINE REVIEW')
    print('Every response below comes from the cells you last ran. No download is needed.')
    print('Part 1 uses simulated data. Part 2 combines published summaries, attributed empirical excerpts, and bibliographic metadata; each supports different claims.')
    print()
    show_response_block('1. Your current settings and source status', {**configuration, 'retrieval_used': RETRIEVAL_USED, 'retrieved_sample_n': len(df), 'annual_count_status': trend_status})
    show_response_block('2. Q1–Q6 and AI-use disclosure', answers)
    show_response_block('3. Provisional innovation canvas', canvas)
    show_response_block('4. Manual search-gap audit', audit)
    for i, check in enumerate(checks, start=1):
        show_response_block(f'5.{i} Original-source check — status is student-reported', check)
    show_response_block('6. Your empirical research-design draft', design)
    show_response_block('7. Actual empirical output and its boundary', empirical_metrics)
    print('Published skill-measure comparison — actual summary values from Part 2')
    display(paper_measure_results)
    print('Controlled simulation — held-out prediction comparison')
    display(scores)
    print('Controlled simulation — randomized estimate')
    display(causal)
    print('Controlled simulation — allocation under the chosen budget')
    display(policy)
    if 'source_comparison' in globals():
        print('Literature retrieval diagnostics — missing access is not zero literature')
        display(source_comparison)
    if len(df):
        print('Primary literature sample — open the sources before accepting method suggestions')
        display(df[['title', 'year', 'venue', 'url']].head(12))
    print('Charts remain visible in Parts 1 and 2; use the relevant chart with these responses.')
    status_fields = {**answers, **canvas, **audit, **{f'SOURCE_{i}_{k}': v for i, check in enumerate(checks, 1) for k, v in check.items()}}
    status = part3_draft_status(status_fields)
    for i, check in enumerate(checks, 1):
        if check['STATUS'] != 'I checked the original':
            print(f'Source {i}: {check["STATUS"]}. This is not a completed student verification.')
    if MANUAL_CHECK_STATUS != 'I performed this search':
        print('Manual audit status: ' + MANUAL_CHECK_STATUS + '. Record only searches you actually performed.')
    print('If you cannot save or download, copy your displayed responses and the required figures into the course-approved submission channel while the session is open.')
    return {'settings': configuration, 'answers': answers, 'innovation_canvas': canvas, 'manual_audit': audit, 'source_checks': checks, 'research_design': design, 'draft_status': status}

PARAMETER_GUIDE = [{'parameter': 'GROUP_LABEL', 'default': 'Group 01', 'definition': 'A short group or individual label', 'allowed': 'Any nonempty text', 'example': 'Group 03', 'expected_effect': 'Changes the saved group label; calculations remain the same.', 'part': 'Part 1'}, {'parameter': 'MEMBER_COUNT', 'default': 2, 'definition': 'Number of participants', 'allowed': '1, 2, or 3', 'example': '1', 'expected_effect': 'Records participation; calculations remain the same.', 'part': 'Part 1'}, {'parameter': 'CONTEXT', 'default': 'A community or setting we want to understand', 'definition': 'The population or setting you intend to investigate', 'allowed': 'Your own short description', 'example': 'Heritage workers in Kunshan', 'expected_effect': 'Records context; it does not apply a geographic or population filter.', 'part': 'Part 1'}, {'parameter': 'PURPOSE', 'default': 'Prediction', 'definition': 'Your intended empirical purpose', 'allowed': 'Measurement; Explanation; Prediction; Causal inference; Optimization; Frontier methods', 'example': 'Causal inference', 'expected_effect': 'Records your purpose; all demonstrations still run and literature is not filtered by purpose.', 'part': 'Part 1'}, {'parameter': 'PROXY_NOISE', 'default': 0.8, 'definition': 'Standard deviation of error added to the simulated need measure', 'allowed': '0.2 to 2.0, in steps of 0.2', 'example': '0.2, then 2.0', 'expected_effect': 'Usually increases proxy scatter and measurement error when raised; downstream results can change.', 'part': 'Part 1'}, {'parameter': 'BUDGET_SHARE', 'default': 0.3, 'definition': 'Share of 600 simulated people who can receive support', 'allowed': '0.1 to 0.8, in steps of 0.1', 'example': '0.2, then 0.5', 'expected_effect': 'Changes allocation capacity from 120 to 300 people in this example. It does not change the randomized experiment.', 'part': 'Part 1'}, {'parameter': 'SEED', 'default': 201, 'definition': 'Seed controlling simulated data, splits, and randomized model components', 'allowed': 'An integer from 0 to 4294967294', 'example': '202', 'expected_effect': 'Changes the simulated sample and random components; hold fixed when comparing another parameter.', 'part': 'Part 1'}, {'parameter': 'TOPIC', 'default': 'education inequality', 'definition': 'Main literature search query', 'allowed': 'Nonempty text, up to 200 characters; English terms work best with the screening rules', 'example': 'cultural heritage', 'expected_effect': 'Changes literature retrieval and charts; the numerical simulation keeps the same generating process.', 'part': 'Part 2'}, {'parameter': 'COMPARISON_TOPIC', 'default': 'AI employment', 'definition': 'Second query for the annual search-count chart', 'allowed': 'Text up to 200 characters; blank disables this comparison', 'example': 'climate adaptation', 'expected_effect': 'Changes only the comparison trend, not a second full literature gallery or simulation.', 'part': 'Part 2'}, {'parameter': 'START_YEAR', 'default': 2016, 'definition': 'First year included in literature retrieval', 'allowed': 'Integer; 1990 <= START_YEAR <= END_YEAR <= 2100', 'example': '2020', 'expected_effect': 'Narrows or expands the literature period; a future year supplies no future observations.', 'part': 'Part 2'}, {'parameter': 'END_YEAR', 'default': 2026, 'definition': 'Last year included in literature retrieval', 'allowed': 'Integer; START_YEAR <= END_YEAR <= 2100', 'example': '2024', 'expected_effect': 'Excludes later records; current-year coverage is incomplete.', 'part': 'Part 2'}, {'parameter': 'MAX_WORKS', 'default': 60, 'definition': 'Maximum records requested from each source', 'allowed': '30, 60, or 100', 'example': '30, then 100', 'expected_effect': 'Changes the retrieval cap; actual counts can be lower, especially for saved samples.', 'part': 'Part 2'}, {'parameter': 'RETRIEVAL_MODE', 'default': 'auto', 'definition': 'How the notebook obtains bibliographic records', 'allowed': 'auto or snapshot', 'example': 'snapshot', 'expected_effect': 'auto tries live sources with fallback; snapshot uses saved OpenAlex records and skips live source/trend queries.', 'part': 'Part 2'}, {'parameter': 'QUESTION_DRAFT', 'default': '', 'definition': 'Your researchable question', 'allowed': 'Your own text; blank is an unassessed draft', 'example': 'How does access to transport relate to attendance?', 'expected_effect': 'Saved as Q1; does not automatically fit a new model.', 'part': 'Part 1'}, {'parameter': 'PURPOSE_EVIDENCE', 'default': '', 'definition': 'Your purpose and the evidence it needs', 'allowed': 'Your own text; blank is an unassessed draft', 'example': 'Prediction; outcomes observed after the prediction date.', 'expected_effect': 'Saved as Q2; states your reasoning.', 'part': 'Part 1'}, {'parameter': 'INITIAL_EXPECTATION', 'default': '', 'definition': 'What you expect before inspecting results', 'allowed': 'Your own text; blank is an unassessed draft', 'example': 'I expect greater measurement noise to weaken prediction.', 'expected_effect': 'Saved as Q3; preserve this initial view when reporting a revision.', 'part': 'Part 1'}, {'parameter': 'AI_USE_STATEMENT', 'default': 'No generative AI is required by this notebook. Document any separately permitted assistance here.', 'definition': 'Disclosure of any instructor-permitted assistance', 'allowed': 'Your own text; blank is an unassessed draft', 'example': 'No generative AI used.', 'expected_effect': 'Saved as your disclosure; initial reasoning and individual reflection remain Human-Only.', 'part': 'Part 1'}, {'parameter': 'SUPPORTED_FINDING', 'default': 'EXAMPLE: In the published two-occupation table in Part 2, Food Batchmakers have the larger embedding-distance change, while Computer Programmers have the larger skill-frequency change. Changing the measure reverses this pairwise ranking. This does not estimate a causal effect of robot adoption.', 'definition': 'One finding tied to an actual output and its boundary', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: In the published two-occupation table in Part 2, Food Batchmakers have the larger embedding-distance change, while Computer Programmers have the larger skill-frequency change. Changing the measure reverses this pairwise ranking. This does not estimate a causal effect of robot adoption.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'REVISED_QUESTION', 'default': 'EXAMPLE: How sensitive is apparent occupational skill change to the measurement rule, and what independent evidence would justify using that measure?', 'definition': 'Your revised question or contribution direction', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: How sensitive is apparent occupational skill change to the measurement rule, and what independent evidence would justify using that measure?', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'NEXT_EVIDENCE', 'default': 'EXAMPLE: A larger permitted sample for the same occupations and years, plus independent task or worker-transition records to validate the measure. Check access and coverage before promising that test.', 'definition': 'The next evidence you need', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: A larger permitted sample for the same occupations and years, plus independent task or worker-transition records to validate the measure. Check access and coverage before promising that test.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'MANUAL_PLATFORM', 'default': 'EXAMPLE: Google Scholar; a student search has not been performed in this example.', 'definition': 'A complementary platform you actually checked', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: Google Scholar; a student search has not been performed in this example.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'MANUAL_QUERY', 'default': 'EXAMPLE: "occupational skill change" "embedding"', 'definition': 'Exact query used on that platform', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: "occupational skill change" "embedding"', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'ADDED_SOURCE', 'default': 'EXAMPLE: Candidate to inspect: https://doi.org/10.1038/s41467-025-67992-y. Compare it with your actual retrieved list before calling it a missed source.', 'definition': 'A relevant source found manually, or a no-addition record', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: Candidate to inspect: https://doi.org/10.1038/s41467-025-67992-y. Compare it with your actual retrieved list before calling it a missed source.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'MISSED_REASON', 'default': 'EXAMPLE: A query about AI employment may miss occupational skill-transformation terminology. Check the actual result list before attributing an omission.', 'definition': 'Why the first query missed it, or what checks found no addition', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: A query about AI employment may miss occupational skill-transformation terminology. Check the actual result list before attributing an omission.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'SEARCH_REVISION', 'default': 'EXAMPLE: Add occupational skill change and skill transformation as search terms; record whether this changes the closest studies and narrows the claim.', 'definition': 'How your query or contribution claim should change', 'allowed': 'Replace the teaching example with your own checked text.', 'example': 'EXAMPLE: Add occupational skill change and skill transformation as search terms; record whether this changes the closest studies and narrows the claim.', 'expected_effect': 'Updates the displayed draft when this input cell and the final inline review are run.', 'part': 'Part 3'}, {'parameter': 'CLOSEST_STUDIES', 'default': 'EXAMPLE: Tong, Wu & Evans, https://doi.org/10.1038/s41467-025-67992-y (Figure 1 and measurement methods); Ludwig, Mullainathan & Rambachan, https://doi.org/10.1146/annurev-economics-120925-105620 (measurement and validation framework). These are starting comparisons; I still need a closer study for my own question.', 'definition': 'Closest studies', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Tong, Wu & Evans, https://doi.org/10.1038/s41467-025-67992-y (Figure 1 and measurement methods); Ludwig, Mullainathan & Rambachan, https://doi.org/10.1146/annurev-economics-120925-105620 (measurement and validation framework). These are starting comparisons; I still need a closer study for my own question.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'CANDIDATE_GAP', 'default': 'EXAMPLE: Test whether an occupational skill-change ranking is robust to the measurement rule. This is a limited validation question, not a claim that the topic has never been studied.', 'definition': 'Candidate gap', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Test whether an occupational skill-change ranking is robust to the measurement rule. This is a limited validation question, not a claim that the topic has never been studied.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'BASELINE_AND_TEST', 'default': 'EXAMPLE: Compare skill-frequency change with embedding-distance change for the same occupations and years. Keep the population and period fixed, then check whether the ranking changes.', 'definition': 'Baseline and test', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Compare skill-frequency change with embedding-distance change for the same occupations and years. Keep the population and period fixed, then check whether the ranking changes.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'DATA_AND_EVALUATION', 'default': 'EXAMPLE: Begin with the published two-occupation summary in Part 2. Compare ranks within each measure. A larger study needs licensed or permitted data and an independent validation criterion; these two rows cannot establish general performance.', 'definition': 'Data and evaluation', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Begin with the published two-occupation summary in Part 2. Compare ranks within each measure. A larger study needs licensed or permitted data and an independent validation criterion; these two rows cannot establish general performance.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'CONTRIBUTIONS', 'default': 'EXAMPLE: Connect an ML representation to an economic measurement question. Explain what a ranking can tell retraining planners and which evidence is still missing before making a policy recommendation.', 'definition': 'Contributions', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Connect an ML representation to an economic measurement question. Explain what a ranking can tell retraining planners and which evidence is still missing before making a policy recommendation.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'PEER_FEEDBACK_REVISION', 'default': 'EXAMPLE: A possible objection is that two selected occupations cannot establish a general pattern. Ask a peer, record their actual objection, and narrow the claim or expand the validation sample.', 'definition': 'Peer feedback revision', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: A possible objection is that two selected occupations cannot establish a general pattern. Ask a peer, record their actual objection, and narrow the claim or expand the validation sample.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'NEXT_WEEK_ACTION', 'default': 'EXAMPLE: Locate two closer studies, check data access, and prepare a one-page measurement-validation plan for PS1. Ask for help selecting an independent criterion.', 'definition': 'Next week action', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Locate two closer studies, check data access, and prepare a one-page measurement-validation plan for PS1. Ask for help selecting an independent criterion.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'MANUAL_DATE', 'default': 'EXAMPLE: Replace with the actual date of your search, YYYY-MM-DD.', 'definition': 'Manual date', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Replace with the actual date of your search, YYYY-MM-DD.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'AUDIT_TERMS', 'default': 'EXAMPLE: occupational skill change; skill transformation; 职业技能变化. Use terms appropriate to your own question.', 'definition': 'Audit terms', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: occupational skill change; skill transformation; 职业技能变化. Use terms appropriate to your own question.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'BACKWARD_CHECK', 'default': "EXAMPLE: Open a close paper's references and record one relevant item, or the checks that found none. Not yet performed by the student.", 'definition': 'Backward check', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': "EXAMPLE: Open a close paper's references and record one relevant item, or the checks that found none. Not yet performed by the student.", 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'FORWARD_CHECK', 'default': "EXAMPLE: Open a close paper's citation trail and record one later study, or the checks that found none. Not yet performed by the student.", 'definition': 'Forward check', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': "EXAMPLE: Open a close paper's citation trail and record one later study, or the checks that found none. Not yet performed by the student.", 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'VERSION_CHECK', 'default': 'EXAMPLE: Compare the DOI record with any preprint and record the version actually read. Not yet performed by the student.', 'definition': 'Version check', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Compare the DOI record with any preprint and record the version actually read. Not yet performed by the student.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_TITLE', 'default': 'EXAMPLE: Tong, Wu & Evans — occupational skill-transformation study.', 'definition': 'Source 1 title', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Tong, Wu & Evans — occupational skill-transformation study.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_URL', 'default': 'EXAMPLE: https://doi.org/10.1038/s41467-025-67992-y', 'definition': 'Source 1 url', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: https://doi.org/10.1038/s41467-025-67992-y', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_CLAIM', 'default': 'EXAMPLE: The two published measures rank the selected occupations differently.', 'definition': 'Source 1 claim', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: The two published measures rank the selected occupations differently.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_SECTION', 'default': 'EXAMPLE: Main text accompanying Figure 1, PDF page 2; measurement methods. Open and inspect the original yourself.', 'definition': 'Source 1 section', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Main text accompanying Figure 1, PDF page 2; measurement methods. Open and inspect the original yourself.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_DECISION', 'default': 'EXAMPLE: Revise an interpretation of robot-caused job loss to a bounded statement about measured skill change.', 'definition': 'Source 1 decision', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Revise an interpretation of robot-caused job loss to a bounded statement about measured skill change.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_1_REASON', 'default': 'EXAMPLE: The displayed summaries concern measurement and rankings; a causal effect needs a separate identification design.', 'definition': 'Source 1 reason', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: The displayed summaries concern measurement and rankings; a causal effect needs a separate identification design.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_TITLE', 'default': 'EXAMPLE: Ludwig, Mullainathan & Rambachan — Large Language Models: An Applied Econometric Framework.', 'definition': 'Source 2 title', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Ludwig, Mullainathan & Rambachan — Large Language Models: An Applied Econometric Framework.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_URL', 'default': 'EXAMPLE: https://doi.org/10.1146/annurev-economics-120925-105620', 'definition': 'Source 2 url', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: https://doi.org/10.1146/annurev-economics-120925-105620', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_CLAIM', 'default': 'EXAMPLE: Machine-produced labels need validation before they are used as social-science measures.', 'definition': 'Source 2 claim', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Machine-produced labels need validation before they are used as social-science measures.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_SECTION', 'default': 'EXAMPLE: Introduction; estimation and validation framework. Record the actual section or page you read.', 'definition': 'Source 2 section', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Introduction; estimation and validation framework. Record the actual section or page you read.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_DECISION', 'default': 'EXAMPLE: Add an independent validation sample to the proposed measurement study.', 'definition': 'Source 2 decision', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Add an independent validation sample to the proposed measurement study.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}, {'parameter': 'SOURCE_2_REASON', 'default': 'EXAMPLE: Fluent or plausible model output does not establish construct validity in the intended population.', 'definition': 'Source 2 reason', 'allowed': 'Your own checked text; teaching examples are labeled EXAMPLE.', 'example': 'EXAMPLE: Fluent or plausible model output does not establish construct validity in the intended population.', 'expected_effect': 'Printed immediately below the input cell and in the final inline review.', 'part': 'Part 3'}]
LEARNING_OBJECTIVE_MAP = [{'part': 1, 'learning_objectives': 'LO2; LO3; LO4; LO5', 'evidence': 'Initial expectation; controlled parameter comparison; interpretation of an actual output'}, {'part': 2, 'learning_objectives': 'LO1; LO4; LO5', 'evidence': 'Two close studies; two original-source checks; one manual search-gap audit'}, {'part': 3, 'learning_objectives': 'LO5; LO6', 'evidence': 'Supported finding; justified revision; presentation; individual reflection; inline investigation record'}]

print('Supplied functions ready. No live literature query has run yet.')
display(pd.DataFrame([{'package':'Python','version':platform.python_version()},
 {'package':'NumPy','version':np.__version__},{'package':'pandas','version':pd.__version__},
 {'package':'Matplotlib','version':matplotlib.__version__},
 {'package':'scikit-learn','version':__import__('sklearn').__version__},
 {'package':'IPython','version':__import__('IPython').__version__}]))
````

### Measurement: what does an observation stand for?

📖 **READ & LEARN.**

**Intuition.** A social concept such as opportunity or need is not automatically a column in a dataset. Measurement makes a concept empirically examinable by specifying observations and a rule for relating them to that concept. What the rule omits matters.

**Mathematics.** A proxy can be represented as $m_i=g(z_i)+\epsilon_i$, where $z_i$ is a construct, $g$ is the measurement rule, and $\epsilon_i$ is error. In this simulation $m_i=z_i+\epsilon_i$, so we can calculate $\mathrm{MSE}_m=n^{-1}\sum_i(m_i-z_i)^2$. Real social data seldom reveal $z_i$ directly, so validation needs a defensible external criterion.

**Pseudocode**
```text
DEFINE the concept and population
SPECIFY observations and a measurement rule
COMPARE with an appropriate validation criterion
INSPECT errors and exclusions
REVISE the rule or narrow the claim
```

The first plot compares a known simulated construct with a noisy proxy. Change **PROXY_NOISE** and rerun to see what changes. A strong correlation alone does not establish construct validity.

Reference: Ludwig, J., Mullainathan, S., & Rambachan, A. (2026). Large Language Models: An Applied Econometric Framework. *Annual Review of Economics*, 18, 283–316. https://doi.org/10.1146/annurev-economics-120925-105620.

### Explanation: which account fits an observed pattern?

📖 **READ & LEARN.**

**Intuition.** Explanation develops an account of how or why a pattern occurs and exposes that account to criticism. A fitted association or feature attribution describes a model's behavior. A causal explanation requires additional assumptions and evidence about mechanisms or interventions.

**Mathematics.** A descriptive model may approximate $\mathbb E[Y\mid X=x]=f_\theta(x)$. Here a linear model uses $\hat y_i=\hat\beta_0+\hat\beta^T x_i$, estimated by minimizing $\sum_i(y_i-\beta_0-\beta^T x_i)^2$. A coefficient depends on the variables, scaling, specification, and sample. Its sign is not automatically the effect of changing a feature.

**Pseudocode**
```text
STATE the pattern and a possible explanation
FIT a transparent descriptive model
INSPECT coefficients and alternative specifications
COMPARE competing accounts
IDENTIFY the additional evidence needed for a causal claim
```

Pesnot Lerousseau, J., & Summerfield, C. (2026). Shared sensitivity to data distribution during learning in humans and transformer networks. Nature Human Behaviour, 10, 601–614. Published online December 22, 2025. https://doi.org/10.1038/s41562-025-02359-3

### Prediction: what will happen to an unseen case?

📖 **READ & LEARN.**

**Intuition.** Prediction concerns outcomes unavailable when a decision must be made. A useful predictive test imitates that information boundary. Predicting a final outcome with a variable observed only afterwards can be numerically impressive and practically unusable.

**Mathematics.** Fit $\hat f$ on training observations and assess $\widehat R_{test}=n_{test}^{-1}\sum_{i\in test}L(y_i,\hat f(x_i))$. We use squared error and report $\mathrm{RMSE}=\sqrt{\widehat R_{test}}$. A mean baseline, linear regression, and a small random forest share the same training/test split. This random split is a teaching design; temporal or geographic deployment may require different splits.

**Pseudocode**
```text
SPECIFY the outcome, prediction time, and available features
SPLIT before model fitting
FIT a simple baseline and candidate models on training data
EVALUATE all models on the same held-out data
CHECK leakage, calibration, and relevance to the intended use
```

Reference: Hollmann, N., et al. (2025). Accurate Predictions on Small Data with a Tabular Foundation Model. *Nature*, 637, 319–326. https://doi.org/10.1038/s41586-024-08328-6.

### Causal inference: what changes because of an intervention?

📖 **READ & LEARN.**

**Intuition.** An outcome can be predictable without being changeable by the proposed action. Causal inquiry asks how outcomes would differ under alternative interventions, not merely how observed groups differ.

**Mathematics.** The average treatment effect is $\tau=\mathbb E[Y(1)-Y(0)]$. Under randomized assignment, consistency, no interference, and positive assignment probabilities, the difference in means estimates this effect. The demonstration uses $\hat\tau=\bar Y_{T=1}-\bar Y_{T=0}$ and an approximate independent-sample standard error. The simulation reveals both potential outcomes to the evaluator; an actual study observes only one per person.

**Pseudocode**
```text
DEFINE treatment, outcome, population, and causal estimand
JUSTIFY identification assumptions and assignment design
ESTIMATE the effect and uncertainty
CHECK the design and its limitations
STATE the population and intervention to which the claim applies
```

Intellectual lineage: Royal Swedish Academy of Sciences (2021), *Answering Causal Questions Using Observational Data*, https://www.nobelprize.org/prizes/economic-sciences/2021/advanced-information/; ACM (2012), Judea Pearl Turing Award announcement, https://www.acm.org/binaries/content/assets/press-releases/2012/march/turing-award-11.pdf. Contemporary application: Brynjolfsson, E., Li, D., & Raymond, L. (2025). Generative AI at Work. *QJE*, 140(2), 889–942. https://doi.org/10.1093/qje/qjae044.

### Optimization: which feasible action serves the objective?

📖 **READ & LEARN.**

**Intuition.** Knowing who has a low expected outcome and knowing who benefits most from help are different forms of knowledge. A decision also needs an objective, constraints, and a view about whose outcomes count.

**Mathematics.** For a binary allocation $a_i$, solve $\max_{a_i\in\{0,1\}}\sum_i a_i\tau_i$ subject to $\sum_i a_i\leq B$. Our comparison uses equal per-person cost and a specified capacity. The known-effect oracle uses true simulated $\tau_i$ as an upper reference, not an available real-world decision rule. Estimating treatment effects, costs, fairness constraints, and policy value would be separate tasks in a deployed setting.

**Pseudocode**
```text
STATE the objective and whose welfare it represents
SPECIFY feasible actions, costs, and constraints
COMPARE a simple policy with justified alternatives
EVALUATE value using an appropriate design
EXAMINE distributional consequences and sensitivity
```

Reference: Athey, S., Keleher, N., & Spiess, J. (2025). Machine Learning Who to Nudge: Causal vs Predictive Targeting in a Field Experiment on Student Financial Aid Renewal. *Journal of Econometrics*, 249, 105945. https://doi.org/10.1016/j.jeconom.2024.105945. See also Athey (2025), *AER*, 115(4), 1059–1099, https://doi.org/10.1257/aer.115.4.1059.

## 1.7 How would you evaluate a frontier method?

📖 **READ & LEARN.**

**Intuition.** Foundation models and research agents can support several empirical purposes. Their novelty does not decide whether they help your question. A fair evaluation holds the task, data access, comparison, and resource budget sufficiently clear to interpret a difference.

**Mathematics.** A tabular in-context model can use a learned conditional predictor $\hat p_\theta(y_*\mid x_*,D_{train})$. A comparison might examine $\Delta R=R(f_{baseline})-R(f_{candidate})$, together with runtime, memory, access conditions, and failure cases. An agent-assisted workflow still needs a human-defined objective and verification criteria.

**Pseudocode**
```text
DEFINE the empirical purpose and baseline
FIX the evaluation data and permitted information
RECORD training, runtime, and resource conditions
COMPARE the new method with the baseline
VERIFY outputs and report failures as well as improvements
```

## 2.1C Deep dive B: measuring skill adaptation in the future of work

📖 **READ & LEARN.** Tong, Wu and Evans (2026), *Nature Communications*, analyze 167 million Lightcast job ads from 2010–2018, covering 721 occupations. A skip-gram word2vec model learns 200-dimensional vectors for 15,182 skills. A job ad supplies a co-occurrence context; an occupation-year vector averages its core skill vectors. The empirical question is who faces larger skill changes when the measure accounts for distance between skills.

**Primary course label: Measurement using representation learning.** The paper describes observational differences and validates skill proximity against worker mobility. It does not identify a robot-adoption causal effect, forecast an unseen robot economy or optimize retraining allocation. Word2vec is an established ML method used here to address a consequential social-science measurement gap.

### How the ML measure works

Let $e_s$ be the learned vector for skill $s$, and $S_{o,t}$ the core skills for occupation $o$ in year $t$. The occupation representation is $z_{o,t}=|S_{o,t}|^{-1}\sum_{s\in S_{o,t}}e_s$. Its change is

$$d_o=1-\frac{z_{o,2010}^{\top}z_{o,2018}}{\lVert z_{o,2010}\rVert\lVert z_{o,2018}\rVert}.$$

Pseudocode: extract skill co-occurrences from ads; learn skill vectors; average core skills by occupation/year; calculate cosine distance; compare the baseline ranking; validate against independent information. The formula agrees with the authors’ MainFigures notebook, section Job Vector Change.

**Innovation question:** in workplaces adopting robots, which new skill requirements do existing measures miss? A proposed study could collect current ads, compare a count baseline with a learned measure and validate against independent assessments of training demands. Establish the closest precedents and what would falsify the proposed advantage. This is a proposed extension, not a result about robot adoption from the source study.

Article: https://doi.org/10.1038/s41467-025-67992-y

Author code and processed data: https://github.com/di-Tong/SkillPaper

Archived release: https://doi.org/10.5281/zenodo.17444902

The raw ads require a Lightcast license. This Week 1 run interprets published summaries and does not retrain embeddings or reanalyse the full proprietary corpus.

## 2.1D Comparing two published measures

The next cell uses four reported values from the main text accompanying Figure 1: skill-frequency change is 7.453 for computer programmers and 2.708 for food batchmakers; holistic embedding distance is 0.017 and 0.082 respectively. These are different scales. Compare occupational rankings within each measure, not raw magnitudes between measures. The holistic distances differ from the approximate skill contributions plotted in Figure 1d, as explained in Supplementary Note 2.2.

**SKILL_CHANGE_MEASURE:** the measure to plot, default `embedding_distance`; allowed `embedding_distance` or `skill_frequency`. Example: switch to `skill_frequency` and rerun. The ranking reverses because the two measures operationalize change differently. You are changing a representation choice, not the source data or an intervention.

**OCCUPATION_TO_HIGHLIGHT:** which occupation is highlighted, default `Food Batchmakers`; allowed `Food Batchmakers` or `Computer Programmers`. Example: select `Computer Programmers`. It changes the bar color only, not any value or ranking.

🔎 **HUMAN CHECK.** Check the four values in the original text, explain why the ranking changes and identify one limitation of treating job-ad skill distance as workers’ actual learning cost. Preserve a candidate new ML task, baseline and validation plan for PS1.

````python
# @title ✏️ STUDENT INPUT — Published skill-measure comparison
SKILL_CHANGE_MEASURE = "embedding_distance" # @param ["embedding_distance", "skill_frequency"]
OCCUPATION_TO_HIGHLIGHT = "Food Batchmakers" # @param ["Food Batchmakers", "Computer Programmers"]
````

````python
# @title ▶️ RUN & OBSERVE — A measurement choice changes the published ranking
assert SKILL_CHANGE_MEASURE in ["embedding_distance","skill_frequency"]
assert OCCUPATION_TO_HIGHLIGHT in ["Food Batchmakers","Computer Programmers"]
paper_measure_results = pd.DataFrame({
    'occupation':['Computer Programmers','Food Batchmakers'],
    'skill_frequency':[7.453,2.708], 'embedding_distance':[0.017,0.082],
    'source_doi':['10.1038/s41467-025-67992-y']*2,
    'source_locator':['Main text accompanying Figure 1, PDF page 2']*2})
paper_measure_results['rank_selected_measure'] = paper_measure_results[SKILL_CHANGE_MEASURE].rank(ascending=False,method='min').astype(int)
paper_measure_results.to_csv(OUT/'paper_measure_comparison.csv',index=False)
MEASUREMENT_SETTINGS_USED = {name:globals()[name] for name in ['SKILL_CHANGE_MEASURE','OCCUPATION_TO_HIGHLIGHT']}
(OUT/'paper_measure_settings.json').write_text(json.dumps({**MEASUREMENT_SETTINGS_USED,
    'status':'Published summary values, not a raw-data replication',
    'source':'https://doi.org/10.1038/s41467-025-67992-y',
    'units':'Different measures have different scales; compare ranks within a measure.'},indent=2))
fig, ax = plt.subplots(figsize=(7,3.6))
ax.bar(paper_measure_results.occupation,paper_measure_results[SKILL_CHANGE_MEASURE],
       color=['#14877D' if o==OCCUPATION_TO_HIGHLIGHT else '#A8B8C8' for o in paper_measure_results.occupation])
ax.set(ylabel='Cosine distance' if SKILL_CHANGE_MEASURE=='embedding_distance' else 'Skill-frequency change score',
       title='Published occupational change: '+SKILL_CHANGE_MEASURE.replace('_',' '))
fig.text(0.5,0.005,'Tong, Wu & Evans, Nature Communications 17, 1237. Published summary, 2010–2018.',ha='center',fontsize=8)
fig.tight_layout(rect=(0,0.04,1,1));fig.savefig(OUT/'paper_measure_comparison.png',dpi=150)
fig.savefig(OUT/'paper_measure_comparison.pdf');display(fig);plt.close(fig)
display(paper_measure_results)
print('Highest measured change:',paper_measure_results.loc[paper_measure_results[SKILL_CHANGE_MEASURE].idxmax(),'occupation'])
print('This compares published measures. It does not identify a causal effect or retrain skill2vec.')
````

## 2.1C Monday feedback and scientific reasoning

📖 **READ & LEARN.** Monday's discussion surfaced three questions: What is new? Which method could support the intended claim? Can we obtain suitable evidence? Your revised design should answer all three before Part 3's Innovation canvas.

**Two routes to originality.** A method upgrade should improve a consequential research task against a baseline. A new economic setting should introduce a changed actor, mechanism or constraint that makes a new empirical question possible. Compare two close studies and say what your study could establish beyond them. A new topic name or a different algorithm alone leaves the contribution unclear. The Xu role taxonomy and Tong skill-measurement study illustrate these routes; proposed student extensions remain hypotheses.

| Scientific approach | What warrants a claim | What a further empirical claim needs |
|---|---|---|
| Deduction or theory | A derivation from stated assumptions | Evidence about whether those assumptions and implications apply |
| Computational simulation | A specified mechanism and verified implementation | Calibration or validation against observations for claims about a real setting |
| Observational empirical analysis | Recorded observations, a design and valid measurement | Validation, alternative explanations and uncertainty appropriate to the research purpose |
| Behavioral experiment | Observed responses under a designed comparison | Valid instruments and design; institutional review and consent when required |

Behavioral research **is empirical**. In this course, an appropriate ML task must contribute to the empirical investigation. Reusing an existing dataset can make a first study feasible. A new survey or experiment can place the contribution in human-data collection and design, with additional time and review requirements. Public data do not automatically settle privacy, terms or review status. Consult the instructor and DKU IRB as appropriate before human-subject research activities.

**Claim check.** Measurement requires construct validation. Explanation examines a pattern and alternatives. Prediction requires held-out cases. Causal inference needs identification of a counterfactual contrast. Optimization needs a defensible objective, constraints and valid inputs. One model score cannot establish all five.

### What each empirical demonstration can establish

▶️ **RUN & OBSERVE.** The next cell produces a figure from actual source records. Interpret its unit and sampling frame before discussing a wider population.

* **FMA:** a descriptive measurement of the music archive's published genre taxonomy and counts. Multiple genre assignments can overlap. It does not train a music classifier or measure aesthetic quality.
* **FinQA:** a descriptive inspection of the first 200 training examples and their arithmetic programs. These are selected training records, so the chart is not held-out model performance or a claim about financial markets.
* **GHCN-Daily:** a small prediction comparison using one weather station. A chronological split keeps future outcomes out of training. The model uses today's temperatures and seasonal features. Persistence predicts the future temperature using today's maximum. Test MAE is in degrees Celsius. A local example does not establish a global climate trend or a policy effect.

🔎 **HUMAN CHECK.** Identify one observed record, the measured variable and its source. Change one applicable setting and explain an actual difference. Then ask which data and validation your own question would require.

````python
# @title ▶️ RUN & OBSERVE — Inspect empirical evidence and its limits
assert EMPIRICAL_DEMO in ['fma','finqa','ghcn']
assert DATASET_ID in [d['id'] for d in DATASET_CATALOG]
assert selected_dataset['id'] == DATASET_ID, 'Catalog selection changed; rerun the dataset-passport cell.'
assert 50 <= MAX_ROWS <= 800 and MAX_ROWS % 50 == 0
assert FMA_VIEW in ['Top-level genres','All genres']
assert HORIZON_DAYS in [1,3,7] and 0.2 <= TEST_SHARE <= 0.4
fig, ax = plt.subplots(figsize=(8.4,4.3))
empirical_metrics = {}
if EMPIRICAL_DEMO == 'fma':
    empirical = pd.read_csv(io.StringIO(EMPIRICAL_SNAPSHOTS['fma_genres.csv']))
    empirical = empirical.head(int(MAX_ROWS)).copy()
    view = empirical.loc[empirical['parent'] == 0] if FMA_VIEW == 'Top-level genres' else empirical
    shown = view.nlargest(8,'#tracks').sort_values('#tracks')
    ax.barh(shown['title'],shown['#tracks'],color='#22749C')
    ax.set(xlabel='Published track-to-genre assignments (may overlap)',title='FMA: archive composition in a selected taxonomy view')
    empirical_metrics={'analysis':'descriptive archive inspection','rows_inspected':len(empirical),'genre_rows_in_view':len(view),'setting':FMA_VIEW,'limit':'Overlapping genre counts are not unique-track counts or population shares.'}
elif EMPIRICAL_DEMO == 'finqa':
    empirical = pd.DataFrame(json.loads(EMPIRICAL_SNAPSHOTS['finqa_training_excerpt.json'])).head(int(MAX_ROWS)).copy()
    empirical['first_operation'] = empirical['program'].str.extract(r'([A-Za-z_]+)\(').fillna('unparsed')
    operations=empirical['first_operation'].value_counts().head(8).sort_values()
    ax.barh(operations.index,operations.values,color='#22749C')
    ax.set(xlabel='Examples in this selected training excerpt',title='FinQA: first arithmetic operation in training records')
    empirical_metrics={'analysis':'descriptive training-record inspection','rows_inspected':len(empirical),'limit':'First training examples, not a random sample or a test evaluation.'}
else:
    from sklearn.linear_model import Ridge
    empirical = pd.read_csv(io.StringIO(EMPIRICAL_SNAPSHOTS['ghcn_central_park_2018_2020.csv']),parse_dates=['date']).head(int(MAX_ROWS)).copy()
    empirical['target_date']=empirical['date']+pd.to_timedelta(HORIZON_DAYS,unit='D')
    targets=empirical[['date','tmax_c']].rename(columns={'date':'target_date','tmax_c':'target_c'})
    model_data=empirical.merge(targets,on='target_date',how='inner').sort_values('date')
    model_data['season_sin']=np.sin(2*np.pi*model_data.date.dt.dayofyear/365.25)
    model_data['season_cos']=np.cos(2*np.pi*model_data.date.dt.dayofyear/365.25)
    split_time=model_data.date.iloc[int(len(model_data)*(1-TEST_SHARE))]
    train=model_data.loc[model_data.target_date < split_time].copy()
    test=model_data.loc[model_data.date >= split_time].copy()
    assert len(train)>15 and len(test)>5 and train.target_date.max()<test.date.min()
    features=['tmax_c','tmin_c','season_sin','season_cos']
    model=Ridge(alpha=1.0).fit(train[features],train.target_c)
    test['ridge_prediction_c']=model.predict(test[features])
    test['persistence_prediction_c']=test['tmax_c']
    ridge_mae=float(np.abs(test.target_c-test.ridge_prediction_c).mean())
    persistence_mae=float(np.abs(test.target_c-test.persistence_prediction_c).mean())
    ax.plot(test.target_date,test.target_c,label='Observed future maximum',color='#22749C')
    ax.plot(test.target_date,test.ridge_prediction_c,label='Ridge',color='#75579B',alpha=.85)
    ax.plot(test.target_date,test.persistence_prediction_c,label='Persistence',color='#A8631E',alpha=.65)
    ax.set(ylabel='Maximum temperature (degrees C)',title=f'Central Park: {HORIZON_DAYS}-day prediction on later dates')
    ax.tick_params(axis='x',rotation=25);ax.legend(fontsize=8)
    test.to_csv(OUT/'empirical_weather_test_predictions.csv',index=False)
    empirical_metrics={'analysis':'chronological held-out prediction','rows_inspected':len(empirical),'train_rows':len(train),'test_rows':len(test),'horizon_days':HORIZON_DAYS,'split_date':str(split_time.date()),'ridge_mae_c':ridge_mae,'persistence_mae_c':persistence_mae,'train_last_target':str(train.target_date.max().date()),'test_first_input':str(test.date.min().date()),'limit':'One station and a small period; no global trend, causal or policy effect.'}
fig.tight_layout();fig.savefig(OUT/'empirical_evidence.png',dpi=160);fig.savefig(OUT/'empirical_evidence.pdf')
display(fig);plt.close(fig);display(empirical.head(4));display(pd.DataFrame(list(empirical_metrics.items()),columns=['Check','Observed result or limit']))
empirical.to_csv(OUT/'empirical_rows_inspected.csv',index=False)
(OUT/'empirical_metrics.json').write_text(json.dumps(empirical_metrics,indent=2))
(OUT/'dataset_catalog.json').write_text(json.dumps(DATASET_CATALOG,indent=2))
(OUT/'snapshot_provenance.json').write_text(json.dumps(SNAPSHOT_PROVENANCE,indent=2))
EMPIRICAL_SETTINGS_USED={k:globals()[k] for k in ['DATASET_ID','EMPIRICAL_DEMO','MAX_ROWS','FMA_VIEW','HORIZON_DAYS','TEST_SHARE']}
print('Compare observed outputs and limits. A more complex model need not beat its baseline.')
````

## 2.6 Compare annual database-query counts

▶️ **RUN & OBSERVE.** If OpenAlex's live grouping service is available, the next cell plots annual counts for `TOPIC` and `COMPARISON_TOPIC`.
````python
# @title ▶️ RUN & OBSERVE — Compare annual database-query counts when available
trend_frames=[];trend_status=[]
for query in dict.fromkeys([TOPIC,COMPARISON_TOPIC]):
    if not query.strip():continue
    trend,status=global_trend(query);trend_status.append(query+': '+status)
    if trend is not None and len(trend):
        trend['query']=query;trend_frames.append(trend)
if trend_frames:
    trends=pd.concat(trend_frames,ignore_index=True)
    trends.to_csv(OUT/'openalex_searchwide_annual_counts.csv',index=False)
    fig,ax=plt.subplots(figsize=(8.4,4.4))
    for i,(q,g) in enumerate(trends.groupby('query')):ax.plot(g.year,g.indexed_matching_works,marker='o',label=q,color=COLORS[i%len(COLORS)])
    ax.set(xlabel='Publication year',ylabel='Indexed works matching the recorded search',title='Database-wide query counts, not a measure of research quality')
    ax.legend();ax.text(.01,-.23,'Query wording, indexing, and partial current-year coverage affect comparison.',transform=ax.transAxes,fontsize=9)
    save_figure(fig,'11_searchwide_annual_trends')
    display(trends.tail(6))
else:
    print('Database-wide trend counts are unavailable in this run. The sample-year plot above remains labeled as a sample.')
print('\n'.join(trend_status))
````
